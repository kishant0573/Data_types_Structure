## 1. what are data structures, and why are they important?

data structures are ways to organize and store data in a computer so that they can be accessed and modified efficiently.  
they help in managing large amounts of data, improving performance, and solving problems effectively.

**example:**
```python
# example of a list and dictionary ( 2 types of data structures)
my_list = [1, 2, 3, 4]  
my_dict = {"name": "rituraj", "age": 25}  

print(my_list)
print(my_dict)
```

## 2. what is the difference between mutable and immutable data types?

- **Mutable:** these data types can be changed after creation. Example: Lists (`[1, 2, 3]` can be modified).  
- **Immutable:** these cannot be changed after creation. xxample: Tuples (`(1, 2, 3)` remains the same).

**example:**
```python
# mutable example: List
mutable_list = [1, 2, 3]
mutable_list.append(4)  
print(mutable_list)  # output: [1, 2, 3, 4]

# immutable example: Tuple
immutable_tuple = (1, 2, 3)
# immutable_tuple[0] = 10  #cause an error
print(immutable_tuple)  # output: (1, 2, 3)
```

## 3. What are the main differences between lists and tuples in Python?

- lists are **mutable**, while tuples are **immutable**.
- lists are slower than tuples because of their dynamic nature.
- tuples are safer when data should not be changed.

**example:**
```python
# list example
my_list = [1, 2, 3]
my_list.append(4)  # lists can be modified
print(my_list)  # output: [1, 2, 3, 4]

# tuple example
my_tuple = (1, 2, 3)
# my_tuple.append(4)  # will cause an error because tuples cannot be modified
print(my_tuple)  # tput: (1, 2, 3)
```

## 4. how do dictionaries store data?

dictionaries store data as **key-value pairs**. each key is unique and maps to a specific value. python uses a **hash table** internally to make lookups fast.

**example:**
```python
# creating a dictionary
student = {"name": "suraj", "age": 25, "city": "india"}

# Accessing values
print(student["name"])  # output: surj
print(student.get("age"))  # utput: 25
```

## 5. why might you use a set instead of a list in Python?

- sets store **only unique elements** (no duplicates).
- they provide **faster membership testing** (`in` operation).
- useful for **mathematical operations** like union and intersection.

**xample:**
```python
# creating a set
numbers = {1, 2, 3, 3, 4, 5}
print(numbers)  # Out: {1, 2, 3, 4, 5}  (removes duplicate 3)

# checking membership
print(2 in numbers)  # output: True
```

## 6. what is a string in Python, and how is it different from a list?

a string is a **sequence of characters** enclosed in quotes (`"hello"`). Unlike lists, strings are **immutable**, meaning they cannot be changed after creation.

**example:**
```python
# string example
text = "hello"
# text[0] = "H"  # this will cause an error since strings are immutable
print(text.upper())  # put: HELLO
```

## 7. how do tuples ensure data integrity in python?

since tuples are **immutable**, they prevent accidental modifications, ensuring the data remains unchanged.

**example:**
```python
# tuple Example
coordinates = (10, 20)
# coordinates[0] = 15  # This will cause an error
print(coordinates)  # Output: (10, 20)
```

## 8. what is a hash table, and how does it relate to dictionaries in Python?

A **hash table** is a data structure that stores **key-value pairs** and allows fast lookups.  
Python **dictionaries** use hash tables to map keys to values efficiently.

**Example:**
```python
# Dictionary (Uses Hash Table)
data = {"apple": 10, "banana": 20, "orange": 30}
print(data["banana"])  # Output: 20
```

## 11. How do sets handle duplicate values in python?

Sets **automatically remove duplicate values**, ensuring all elements are **unique**.

**Example:**
```python
# Set Example

print(unique_numbers)  # Output: {1, 2, 3, 4}
```

## 12. what is a nested dictionary, and give an example of its use case?

A **nested dictionary** is a dictionary inside another dictionary. This is useful for **storing structured data**.

**example:**
```python
# Nested Dic example
student = {
    "name": "naresh ",
    "grades": {"math": 90, "science": 85}
}
print(student["grades"]["math"])  # utput: 90
```

## 13. describe the time complexity of accessing elements in a dictionary.

accessing an element in a dictionary has an **average time complexity of O(1)** due to hash tables.

## 14. Explain the difference between a list and a dictionary in terms of data retrieval.

- **lists** retrieve values by **index** (`list[0]`).
- **dictionaries** retrieve values by **key** (`dict["name"]`), making lookups faster and more meaningful.

**example:**
```python
# list ret
my_list = ["apple", "banana", "cherry"]
print(my_list[1])  # Output: banana

# dictionary retrieval
my_dict = {"fruit1": "apple", "fruit2": "banana"}
print(my_dict["fruit2"])  # Output: banana
```
